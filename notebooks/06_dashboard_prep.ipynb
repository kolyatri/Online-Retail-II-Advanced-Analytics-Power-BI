**Dashboard Preparation**

Goal

Prepare a clear dashboard structure and narrative so that:
- every page has a purpose
- every KPI answers a business question
- no visual exists “just because”

No DAX changes unless a real gap is discovered.
No design polish yet.

**Define the Dashboard Audience (Lock First)**

Target audience:
- Business stakeholders
- Hiring managers / interviewers
- Non-technical users

Implications:
- Fewer KPIs per page
- Clear naming
- Strong hierarchy (overview → drill-down)
