```python
import pandas as pd
import numpy as np

dtypes = {
    "Invoice": "string",
    "StockCode": "string",
    "Description": "string",
    "Quantity": "int32",
    "Price": "float32",
    "Customer ID": "Int64",
    "Country": "category"
}

df = pd.read_csv(
    "../data/raw/online_retail_II.csv",
    encoding="ISO-8859-1",
    dtype=dtypes,
    parse_dates=["InvoiceDate"]
)

# print(df.sample(10))

# The dataset contains non-UTF characters in product descriptions.
# Using a stable encoding avoids silent text corruption.

# df.shape

# df.info()

# Focus points:

# - InvoiceDate should be datetime
# - Quantity and Price must be numeric
# - CustomerID may contain missing values

# These directly affect:

# - time-based analysis
# - revenue calculation
# - customer segmentation

# df.describe()

# Initial red flags to watch for:

# - negative Quantity
# - zero or negative Price
# - extreme values that may distort averages

# df["InvoiceDate"].agg(["min", "max"])

# Business relevance:

# - Confirms analysis period

# Needed later for:

# - trend analysis
# - cohort analysis
# - retention windows

# df.isna().mean().sort_values(ascending=False)

# Missing CustomerID impacts:

# - customer ranking
# - repeat purchase analysis

# Other missing fields may affect product-level insights

# df.duplicated().sum()

# Important note:

# - Duplicate invoice lines are not always errors.
# - We only remove duplicates if they are fully identical rows.

pd.DataFrame({
    "negative_quantity": [(df["Quantity"] < 0).mean()],
    "zero_or_negative_price": [(df["Price"] <= 0).mean()],
    "missing_customer_id": [df["Customer ID"].isna().mean()]
})

# Why this matters:

# - Negative quantities may indicate returns
# - Zero prices may indicate data entry errors or promotions
# - Missing customers limit customer-level KPIs

df.to_parquet("../data/processed/online_retail_after_ingestion.parquet")

# save cleaned dataframe to share among different ipynb files

```
